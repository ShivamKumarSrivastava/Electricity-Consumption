**TRAINING THE MODEL**

```python
from sklearn.model_selection import train_test_split
```

```python

from sklearn.metrics import r2_score
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
```

```python
import joblib
```

```python
# model 1 for zone 1

model1 = RandomForestRegressor() 

x1 = df.drop(['PowerConsumption_Zone1'],axis=1)
y1 = df['PowerConsumption_Zone1']

x1_train,x1_test,y1_train,y1_test = train_test_split(x1,y1,test_size=0.3,random_state=42)

model1.fit(x1_train,y1_train)
model1_preds = model1.predict(x1_test)

print(f"{r2_score(model1_preds,y1_test)}")
```

```python
# model 2 for zone 2

model2 = RandomForestRegressor() 


x2 = df.drop(['PowerConsumption_Zone2'],axis=1)
y2 = df['PowerConsumption_Zone2']

x2_train,x2_test,y2_train,y2_test = train_test_split(x2,y2,test_size=0.3,random_state=42)

model2.fit(x2_train,y2_train)
model2_preds = model2.predict(x2_test)

print(f"{r2_score(model2_preds,y2_test)}")
```

```python
# model 2 for zone 2

model3 = RandomForestRegressor() 

x3 = df.drop(['PowerConsumption_Zone3'],axis=1)
y3 = df['PowerConsumption_Zone3']

x3_train,x3_test,y3_train,y3_test = train_test_split(x3,y3,test_size=0.3,random_state=42)

model3.fit(x3_train,y3_train)
model3_preds = model3.predict(x3_test)

print(f"{r2_score(model3_preds,y3_test)}")
```

```python
# Saving the model

joblib.dump(model1,"zone1_model.pkl")

joblib.dump(model2,"zone2_model.pkl")

joblib.dump(model3,"zone3_model.pkl")
```
